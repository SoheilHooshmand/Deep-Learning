```python
"""
Bidirectional Recurrent Neural Networks (BRNNs) are an advanced form of RNNs that process sequential data in both forward and backward directions.
This allows the network to use both past and future context, improving understanding and prediction accuracy.
"""
```

```python
# Loading and Preprocessing Data


import warnings
warnings.filterwarnings('ignore')
from keras.datasets import imdb
import numpy as np
from keras.preprocessing.sequence import pad_sequences

features = 2000  
max_len = 50     

# if steel have not download dataset uncomment and use this 
# (X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=features)

with np.load('data/imdb.npz', allow_pickle=True) as f:
    X_train, y_train = f['x_train'], f['y_train']
    X_test, y_test = f['x_test'], f['y_test']

def restrict_words(sequences, num_words):
    return [[w if w < num_words else 2 for w in seq] for seq in sequences]

X_train = restrict_words(X_train, features)
X_test = restrict_words(X_test, features)

X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)
```

```python
# Defining the Model Architecture

from keras.models import Sequential
from keras.layers import Embedding, Bidirectional, SimpleRNN, Dense

embedding_dim = 128  
hidden_units = 64    

model = Sequential()

model.add(Embedding(features, embedding_dim, input_length=max_len))

model.add(Bidirectional(SimpleRNN(hidden_units)))

model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
#  Training the Model

batch_size = 32
epochs = 5

model.fit(X_train, y_train,
          batch_size=batch_size,
          epochs=epochs,
          validation_data=(X_test, y_test))
```

```python
#  Evaluating the Model

loss, accuracy = model.evaluate(X_test, y_test)

print('Test accuracy:', accuracy)
```

```python
# Predicting on Test Data

from sklearn.metrics import classification_report

y_pred = model.predict(X_test)

y_pred = (y_pred > 0.5)

print(classification_report(y_test, y_pred, target_names=['Negative', 'Positive']))
```
