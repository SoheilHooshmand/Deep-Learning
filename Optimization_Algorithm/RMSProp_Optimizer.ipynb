```python
"""
RMSProp is an adaptive optimization algorithm that improves training speed and stability by adjusting the learning rate for each parameter based on recent gradients.
"""
"""
AdaGrad’s main problem is that it keeps accumulating the squared gradients from **all previous iterations**. As training continues,
this accumulated value keeps growing, which causes the effective learning rate to continuously decrease. Eventually, the learning rate can become extremely small,
making the model learn very slowly or even stop making meaningful progress.

RMSProp solves this problem by using an **exponential moving average of the squared gradients** instead of accumulating all of them permanently.
This means recent gradients have more influence, while older gradients gradually lose their effect. As a result, the learning rate can adapt to the current gradients
without continuously shrinking toward zero, making RMSProp more stable and effective for long training processes.
"""
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt
```

```python
# Loading and Preprocessing Dataset



# if have not downloaded data uncomment and use this
# (x_train, y_train), (x_test, y_test) = mnist.load_data()

# x_train = x_train.astype('float32') / 255.0
# x_test = x_test.astype('float32') / 255.0
# y_train = to_categorical(y_train, 10)
# y_test = to_categorical(y_test, 10)

import numpy as np
import struct
from tensorflow.keras.utils import to_categorical

def load_mnist_images(filename):
    with open(filename, 'rb') as f:
        magic, num, rows, cols = struct.unpack('>IIII', f.read(16))
        images = np.frombuffer(f.read(), dtype=np.uint8).reshape(num, rows, cols)
    return images

def load_mnist_labels(filename):
    with open(filename, 'rb') as f:
        magic, num = struct.unpack('>II', f.read(8))
        labels = np.frombuffer(f.read(), dtype=np.uint8)
    return labels

base_path = 'data/MNIST/raw/'

x_train = load_mnist_images(base_path + 'train-images-idx3-ubyte')
y_train = load_mnist_labels(base_path + 'train-labels-idx1-ubyte')
x_test  = load_mnist_images(base_path + 't10k-images-idx3-ubyte')
y_test  = load_mnist_labels(base_path + 't10k-labels-idx1-ubyte')

#  normalization and one_hot
x_train = x_train.astype('float32') / 255.0
x_test  = x_test.astype('float32') / 255.0

y_train = to_categorical(y_train, 10)
y_test  = to_categorical(y_test, 10)
```

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
# Compiling the Model

model.compile(optimizer=tf.keras.optimizers.RMSprop(learning_rate=0.001, rho=0.9),
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# Training the Model

history = model.fit(x_train, y_train, epochs=10,
                    batch_size=32, validation_split=0.2)
```
