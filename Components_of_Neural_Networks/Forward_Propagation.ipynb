```python
"""
Forward propagation is the process where input data passes through each layer of a neural network to produce an output.
It transforms raw inputs into predictions using weights, biases and activation functions.
"""
```

```python
import numpy as np
import pandas as pd
```

```python
# Create Sample Dataset

data = {'cgpa': [8.5, 9.2, 7.8], 'profile_score': [85, 92, 78], 'lpa': [10, 12, 8]}
df = pd.DataFrame(data)
X = df[['cgpa', 'profile_score']].values
```

```python
# Initialize Parameters

def initialize_parameters():
    np.random.seed(1)
    W = np.random.randn(2, 1) * 0.01
    b = np.zeros((1, 1))
    return W, b
```

```python
# Define Forward Propagation

def forward_propagation(X, W, b):
    Z = np.dot(X, W) + b
    A = 1 / (1 + np.exp(-Z))  
    return A
```

```python
# Execute Forward Propagation

W, b = initialize_parameters()
A = forward_propagation(X, W, b)
print("Final Output:", A)
```
